```python
import random

# Initial number of balls
red = 3
blue = 4
black = 2

# Roll a six-sided die
dice = random.randint(1, 6)

# Add a ball depending on the dice result
if dice in [2, 3, 5]:
    black += 1

elif dice == 6:
    red += 1

else:
    blue += 1

# Create the urn with all the balls
urn = []

urn += ["red"] * red
urn += ["blue"] * blue
urn += ["black"] * black

# Draw a random ball
drawn_ball = random.choice(urn)

print("Dice result:", dice)
print("Urn:", urn)
print("Drawn ball:", drawn_ball)
```

```python
# Number of simulations
n = 100000

red_count = 0

for i in range(n):

    red = 3
    blue = 4
    black = 2

    # Roll the dice
    dice = random.randint(1, 6)

    if dice in [2, 3, 5]:
        black += 1

    elif dice == 6:
        red += 1

    else:
        blue += 1

    # Create the urn
    urn = []

    urn += ["red"] * red
    urn += ["blue"] * blue
    urn += ["black"] * black

    # Draw a ball
    drawn_ball = random.choice(urn)

    if drawn_ball == "red":
        red_count += 1

# Estimate the probability
probability = red_count / n

print("Number of simulations:", n)
print("Red balls drawn:", red_count)
print("Estimated probability:", probability)
```


## c) Theoretical Probability Calculation

There are three possible cases when rolling the die:

| Dice result | Probability | Probability of drawing red |
|---|---|---|
| 2, 3, 5 (prime numbers) | 3/6 | 3/10 |
| 6 | 1/6 | 4/10 |
| 1, 4 | 2/6 | 3/10 |

We apply the **law of total probability**:

$$
P(\text{red}) = \frac{3}{6}\cdot\frac{3}{10} + \frac{1}{6}\cdot\frac{4}{10} + \frac{2}{6}\cdot\frac{3}{10}
$$

Therefore, the theoretical probability is:

$$
\boxed{P(\text{red}) = \frac{19}{60} \approx 0.316667}
$$


```python
# Calculate the theoretical probability
p_prime = 3 / 6
p_six = 1 / 6
p_other = 2 / 6

p_red_prime = 3 / 10
p_red_six = 4 / 10
p_red_other = 3 / 10

theoretical_probability = (
    p_prime * p_red_prime
    + p_six * p_red_six
    + p_other * p_red_other
)

# Compare with the simulated probability
difference = abs(probability - theoretical_probability)

print("Estimated probability:", probability)
print("Theoretical probability:", theoretical_probability)
print("Absolute difference:", difference)
```


### Conclusion

The theoretical probability of drawing a red ball is approximately **0.31667 (31.67%)**.

After 100,000 simulations, the estimated probability was **0.31585 (31.585%)**.

The absolute difference between both results is approximately **0.00082**, which is very small.

Therefore, the simulation agrees with the theoretical calculation. The small difference is due to the randomness of the experiment.
