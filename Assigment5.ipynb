### a) Initialization

We need to prove the following formula:

$$
\boxed{\alpha_1(j)=\pi_j b_j(o_1)}
$$

The symbols mean:

- $\alpha_1(j)$: The joint probability of starting in state $j$ and observing the first grade.
- $\pi_j$: The initial probability of state $j$.
- $b_j(o_1)$: The probability of observing the first grade given state $j$.
- $o_1$: The first observed grade.
- $j$: A particular hidden state, such as Difficult.

We use the multiplication rule of probability:

$$
P(A,B)=P(A)\cdot P(B\mid A)
$$

This rule calculates the probability of two events happening together.

In our example:

- Event A: The first exam is Difficult.
- Event B: The student obtains VG.

Therefore:

$$
P(Difficult,VG)=P(Difficult)\cdot P(VG\mid Difficult)
$$


#### Numerical example

From the previous laboratory, we know:

$$
P(Difficult)=\frac13
$$

$$
P(VG\mid Difficult)=0.1
$$

Therefore:

$$
\alpha_1(Difficult)=\frac13\cdot0.1
$$

$$
\boxed{\alpha_1(Difficult)=0.03333}
$$


#### Mathematical proof

By definition, the Forward variable at time 1 is:

$$
\alpha_1(j)=P(O_1=o_1,X_1=j)
$$

Using the multiplication rule of probability:

$$
P(O_1=o_1,X_1=j)
=
P(X_1=j)P(O_1=o_1\mid X_1=j)
$$

By definition of the initial distribution:

$$
P(X_1=j)=\pi_j
$$

And by definition of the emission probabilities:

$$
P(O_1=o_1\mid X_1=j)=b_j(o_1)
$$

Substituting these expressions:

$$
\boxed{\alpha_1(j)=\pi_j b_j(o_1)}
$$

Therefore, the initialization formula is proved.

### b) Recursion

We need to prove the following formula:

$$
\boxed{\alpha_t(j)=b_j(o_t)\sum_i\alpha_{t-1}(i)a_{ij}}
$$

#### Meaning of the symbols

- $t$: The current time step (current exam).
- $j$: The current hidden state.
- $i$: A possible previous hidden state.
- $\alpha_{t-1}(i)$: The joint probability of observing all previous grades and ending in state $i$.
- $a_{ij}$: The probability of transitioning from state $i$ to state $j$.
- $b_j(o_t)$: The probability of observing the current grade in state $j$.
- $\sum_i$: The sum over all possible previous hidden states.

#### Mathematical proof

By definition, the Forward variable is:

$$
\alpha_t(j)=P(O_1=o_1,\ldots,O_t=o_t,X_t=j)
$$

**Step 1: Apply the law of total probability.**

We introduce all possible previous hidden states $X_{t-1}=i$:

$$
\alpha_t(j)
=
\sum_i P(O_1=o_1,\ldots,O_t=o_t,X_{t-1}=i,X_t=j)
$$

We sum over $i$ because the previous state could be any hidden state.

**Step 2: Use the multiplication rule of probability.**

We separate the probability into three parts:

$$
\alpha_t(j)=\sum_i
\underbrace{P(O_1=o_1,\ldots,O_{t-1}=o_{t-1},X_{t-1}=i)}_{\text{1. Past}}
\cdot
\underbrace{P(X_t=j\mid X_{t-1}=i)}_{\text{2. Markov property}}
\cdot
\underbrace{P(O_t=o_t\mid X_t=j)}_{\text{3. Conditional independence}}
$$


**Step 3: Replace each probability with its definition.**

The first part is the previous Forward probability:

$$
P(O_1=o_1,\ldots,O_{t-1}=o_{t-1},X_{t-1}=i)
=
\alpha_{t-1}(i)
$$

The second part is the transition probability:

$$
P(X_t=j\mid X_{t-1}=i)=a_{ij}
$$

The third part is the emission probability:

$$
P(O_t=o_t\mid X_t=j)=b_j(o_t)
$$

Substituting:

$$
\alpha_t(j)=\sum_i\alpha_{t-1}(i)a_{ij}b_j(o_t)
$$

**Step 4: Move the emission probability outside the sum.**

Since $b_j(o_t)$ does not depend on the previous state $i$, it can be moved outside the summation:

$$
\boxed{\alpha_t(j)=b_j(o_t)\sum_i\alpha_{t-1}(i)a_{ij}}
$$

Therefore, the recursion formula is proved.


### c) Termination

We need to prove the following formula:

$$
\boxed{P(O_1=o_1,\ldots,O_T=o_T)=\sum_j\alpha_T(j)}
$$

#### Meaning of the symbols

- $T$: The final time step (last exam).
- $j$: A possible hidden state at the final time step.
- $\alpha_T(j)$: The joint probability of observing the entire sequence and ending in state $j$.
- $\sum_j$: The sum over all possible final hidden states.

At the last exam, the hidden state must be one of the possible difficulties:

- Difficult
- Medium
- Easy

We do not know which state was the final one, so we must consider all of them.

For our example with 12 exams:

$$
P(o_1,\ldots,o_{12})=
\alpha_{12}(Difficult)+
\alpha_{12}(Medium)+
\alpha_{12}(Easy)
$$

#### Mathematical proof

**Step 1: Start with the probability of the complete observation sequence.**

$$
P(O_1=o_1,\ldots,O_T=o_T)
$$

**Step 2: Apply the law of total probability.**

We consider every possible final hidden state $X_T=j$:

$$
P(O_1=o_1,\ldots,O_T=o_T)
=
\sum_j P(O_1=o_1,\ldots,O_T=o_T,X_T=j)
$$

**Step 3: Use the definition of the Forward variable.**

By definition:

$$
\alpha_T(j)=P(O_1=o_1,\ldots,O_T=o_T,X_T=j)
$$

Substituting this into the previous expression:

$$
\boxed{P(O_1=o_1,\ldots,O_T=o_T)=\sum_j\alpha_T(j)}
$$

Therefore, the termination formula is proved.

#### Conclusion

The total probability of the observed sequence is obtained by summing the final Forward probabilities over all possible hidden states.

Forward considers every possible path, while Viterbi selects only the most probable complete path.

## Exercise 2 — Viterbi Algorithm

The objective of this exercise is to implement the Viterbi algorithm manually to find the most probable sequence of hidden states for 12 observed exam grades.

Unlike the Forward algorithm, which sums the probabilities of all possible paths, Viterbi selects the most probable path.

### a) Define the HMM parameters

- D: Difficult
- M: Medium
- E: Easy

We have four possible observations representing grades:

- VG: Very Good

- U: Unsatisfactory

**Initial probabilities**

All three states have the same initial probability:

$$
\pi=\left[\frac13,\frac13,\frac13\right]
$$

**Transition matrix**

The transition matrix $A$ represents the probability of moving from one hidden state to another.

$$
A=
\begin{pmatrix}
0 & 0.5 & 0.5\\
0.5 & 0.25 & 0.25\\
0.5 & 0.25 & 0.25
\end{pmatrix}
$$

Rows and columns follow the order D, M, E.

**Emission matrix**

The emission matrix $B$ represents the probability of observing a grade given the hidden state.

$$
B=
\begin{pmatrix}
0.1 & 0.2 & 0.4 & 0.3\\
0.15 & 0.25 & 0.5 & 0.1\\
0.2 & 0.3 & 0.4 & 0.1
\end{pmatrix}
$$

Rows follow the order D, M, E.

Columns follow the order VG, G, S, U.

Each row of both matrices must sum to 1 because it represents a probability distribution.

### b) Viterbi Initialization

We need to define the Viterbi variable $\delta_t(j)$ and compute its initial values for each hidden state.

#### Definition of the Viterbi variable

$$
\boxed{
\delta_t(j)=
\max_{x_1,\ldots,x_{t-1}}
P(X_1=x_1,\ldots,X_{t-1}=x_{t-1},X_t=j,O_1=o_1,\ldots,O_t=o_t)
}
$$

#### Compute δ1 (j) = πjbj (o1) for each state j.

At the first time step, there are no previous states or transitions to consider.

Therefore:

$$
\boxed{\delta_1(j)=\pi_j b_j(o_1)}
$$

Where:

- $\delta_1(j)$: The joint probability of starting in state $j$ and observing the first grade.
- $\pi_j$: The initial probability of state $j$.
- $b_j(o_1)$: The emission probability of the first observation given state $j$.
- $o_1$: The first observed grade.

\delta_1(E)=\frac13\cdot0.2=0.066667
$$

Therefore:

$$
\boxed{\delta_1=[0.033333,\ 0.05,\ 0.066667]}
$$

### c) Viterbi Recursion and Backpointers

- $\delta_t(j)$: The maximum joint probability of a state sequence ending in state $j$ at time $t$, together with the observations.
- $\psi_t(j)$: The previous hidden state that produces this maximum probability.

#### Recursion formula

$$
\boxed{\delta_t(j)=b_j(o_t)\max_i[\delta_{t-1}(i)a_{ij}]}
$$

#### Backpointers

We also need to remember which previous state produced the maximum probability.

For this, we define:

$$
\boxed{\psi_t(j)=\arg\max_i[\delta_{t-1}(i)a_{ij}]}
$$

The difference between max and argmax is:

- max returns the highest probability.
- argmax returns the previous state that produced that probability.

We store these values in two tables:

- The delta table stores the maximum probabilities.
- The psi table stores the backpointers needed to reconstruct the most probable hidden-state sequence.

```python

import numpy as np

# Hidden states: Difficult, Medium, Easy
states = ["D", "M", "E"]

# Possible observed grades
grades = ["VG", "G", "S", "U"]

# Initial state probabilities
pi = np.array([1/3, 1/3, 1/3])

# Transition probabilities
# Rows: previous state; columns: next state
transition = np.array([
    [0.0, 0.5, 0.5],
    [0.5, 0.25, 0.25],
    [0.5, 0.25, 0.25]
])

# Emission probabilities
# Rows: D, M, E; columns: VG, G, S, U
emission = np.array([
    [0.1, 0.2, 0.4, 0.3],
    [0.15, 0.25, 0.5, 0.1],
    [0.2, 0.3, 0.4, 0.1]
])

# Observed sequence of 12 grades
observations = [
    "VG", "VG", "S", "G", "G", "S",
    "G", "G", "U", "G", "G", "S"
]

# Check that all probability distributions sum to 1
print("Initial probabilities:", pi.sum())
print("Transition row sums:", transition.sum(axis=1))
print("Emission row sums:", emission.sum(axis=1))
```

```python
# Number of observations and hidden states
T = len(observations)
N = len(states)

# Tables for Viterbi probabilities and backpointers
delta = np.zeros((T, N)) # score table 
psi = np.zeros((T, N), dtype=int) # indices of the best preceding states

# Initialization
first_grade_index = grades.index(observations[0])

for j in range(N):
    delta[0, j] = pi[j] * emission[j, first_grade_index]



# Recursion
for t in range(1, T):
    # Get the index of the current observed grade
    grade_index = grades.index(observations[t])

    for j in range(N):
        # Calculate probabilities from all previous states
        candidates = []
        for i in range(N):
            # δ_(t-1)(i) * a_ij
            probability = delta[t - 1][i] * transition[i][j]
            candidates.append(probability)

        # Store the maximum probability
        # δ_t(j) = b_j(o_t) * max_i[δ_(t-1)(i) * a_ij]
        delta[t, j] = emission[j, grade_index] * np.max(candidates)

        # Store the previous state that produced the maximum
        # ψ_t(j) = argmax_i[δ_(t-1)(i) * a_ij]
        psi[t, j] = np.argmax(candidates)


# Display the two tables
print("Delta table:")
print(delta)

print("\nPsi table:")
print(psi)
```

```python
# Display delta values for the first two observations
print("Delta 1:", delta[0])
print("Delta 2:", delta[1])
```

```python
# Backtracking

# Find the most probable final state
last_state = np.argmax(delta[T - 1])

# Start the path with the final state
path = [last_state]

# Go backwards through the previous exams
for t in reversed(range(1, T)):
    # Get the current state (current difficulty level) 
    current_state = path[0]

    # Find the best previous state using psi
    previous_state = psi[t][current_state]  

    # Add the previous state to the path
    path.insert(0, previous_state)

# Calculate the joint probability
best_probability = np.max(delta[T - 1])

# Convert indices into state names
best_sequence = [states[i] for i in path]

print("Most probable sequence:", best_sequence)
print("Joint probability:", best_probability)
```

### f) Interpretation

#### Why does Viterbi find the most probable complete sequence?

The Viterbi algorithm finds the most probable complete sequence of hidden states because it considers the transition probabilities between consecutive states and maximizes the joint probability of the entire path and the observations.

It does not select the most probable hidden state independently at each time step.

For example, in our model, the transition probability from Difficult to Difficult is zero:

$$
P(D \rightarrow D)=0
$$

Therefore, a sequence containing two consecutive Difficult states is impossible, even if Difficult were the most probable state for those tests individually.

Viterbi finds the sequence that maximizes:

$$
\boxed{
x^*=\arg\max_{x_1,\ldots,x_{12}}
P(x_1,\ldots,x_{12},o_1,\ldots,o_{12})
}
$$

#### Joint probability vs. conditional probability

The score obtained in part e is:

$$
\boxed{P(x^*,o)=2.109375\times10^{-11}}
$$

This is a **joint probability**: it represents the probability of the most probable hidden-state sequence occurring together with the observed grades.

It is not the conditional probability of that path given the observations.

The conditional probability would be:

$$
\boxed{
P(x^*\mid o)=\frac{P(x^*,o)}{P(o)}
}
$$

where $P(o)$ is the total probability of the observed grade sequence.

Therefore, the Viterbi score is a joint probability, not a posterior probability.
